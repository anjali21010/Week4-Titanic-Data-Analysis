```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Project output folder
os.makedirs("outputs", exist_ok=True)

# Load Titanic dataset
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

print("TITANIC DATA ANALYSIS PROJECT")
print("-" * 35)
print("Training dataset shape:", train.shape)
print("Testing dataset shape:", test.shape)

# Display first 5 rows
train.head()
```

```python
# Titanic Survival Analysis

print("TOTAL PASSENGERS:", len(train))

print("\nSURVIVAL COUNTS")
print(train["Survived"].value_counts().rename(
    index={0: "Did not survive", 1: "Survived"}
))

print("\nOVERALL SURVIVAL RATE")
print(round(train["Survived"].mean() * 100, 2), "%")

print("\nSURVIVAL RATE BY GENDER")
print((train.groupby("Sex")["Survived"].mean() * 100).round(2))

print("\nSURVIVAL RATE BY PASSENGER CLASS")
print((train.groupby("Pclass")["Survived"].mean() * 100).round(2))
```

```python
# Graph 1: Overall survival count
survival_counts = train["Survived"].value_counts().sort_index()

survival_counts.plot(kind="bar", rot=0)
plt.title("Titanic Passenger Survival Count")
plt.xlabel("Survival (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()


# Graph 2: Survival rate by gender
gender_survival = train.groupby("Sex")["Survived"].mean() * 100

gender_survival.plot(kind="bar", rot=0)
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")
plt.show()


# Graph 3: Survival rate by passenger class
class_survival = train.groupby("Pclass")["Survived"].mean() * 100

class_survival.plot(kind="bar", rot=0)
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.show()
```
